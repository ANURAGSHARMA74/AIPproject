```python
import pytesseract
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
from collections import Counter
from google.colab import files
import re

pytesseract.pytesseract.tesseract_cmd = r'/usr/bin/tesseract'

print("Upload one or more images")
uploaded = files.upload()

for file in uploaded.keys():

    if file.lower().endswith((".jpg", ".png", ".jpeg")):

        print("\n============================")
        print(f"Processing: {file}")
        print("============================")

        # -----------------------------
        # Load Image
        # -----------------------------
        img = Image.open(file)

        plt.imshow(img)
        plt.title("Original Image")
        plt.axis("off")
        plt.show()

        # -----------------------------
        # Preprocessing
        # -----------------------------
        gray = img.convert("L")

        plt.imshow(gray, cmap="gray")
        plt.title("Grayscale Image")
        plt.axis("off")
        plt.show()

        # -----------------------------
        # OCR Extraction
        # -----------------------------
        ocr_text = pytesseract.image_to_string(gray)

        print("\n--- OCR Extracted Text ---\n")
        print(ocr_text)

        # -----------------------------
        # Word Statistics
        # -----------------------------
        words = ocr_text.split()

        print("\n--- Word Statistics ---")
        print("Total Words:", len(words))
        print("Unique Words:", len(set(words)))

        # -----------------------------
        # Dominant Colors
        # -----------------------------
        img_rgb = img.convert("RGB")
        pixels = np.array(img_rgb).reshape(-1, 3)
        color_counts = Counter(map(tuple, pixels))
        dominant_colors = color_counts.most_common(3)

        print("\n--- Dominant Colors (RGB) ---")
        for color, count in dominant_colors:
            print(color)

        # -----------------------------
        # Document Classification
        # -----------------------------
        text_lower = ocr_text.lower()

        if "total" in text_lower or "amount" in text_lower or "receipt" in text_lower:
            doc_type = "Receipt / Bill"
        else:
            doc_type = "General Document"

        print("\n--- Document Classification ---")
        print("Detected Type:", doc_type)

        # -----------------------------
        #  NEW: Extract Total Amount
        # -----------------------------
        amounts = re.findall(r'\d+\.\d{2}', ocr_text)

        if amounts:
            total_amount = max(amounts)
        else:
            total_amount = "Not Found"

        # -----------------------------
        #  NEW: Extract Date
        # -----------------------------
        dates = re.findall(r'\d{2}/\d{2}/\d{4}', ocr_text)

        if dates:
            date = dates[0]
        else:
            date = "Not Found"

        # -----------------------------
        #  NEW: Category Detection
        # -----------------------------
        if any(word in text_lower for word in ["restaurant", "food", "cafe"]):
            category = "Food"
        elif any(word in text_lower for word in ["mall", "shopping", "store"]):
            category = "Shopping"
        elif any(word in text_lower for word in ["fuel", "petrol", "diesel"]):
            category = "Travel"
        elif any(word in text_lower for word in ["hospital", "medical", "pharmacy"]):
            category = "Medical"
        else:
            category = "Other"

        # -----------------------------
        # FINAL OUTPUT
        # -----------------------------
        print("\n---  Extracted Receipt Information ---")
        print("Total Amount:", total_amount)
        print("Date:", date)
        print("Category:", category)

        print("\n Image Analysis Completed")
```
